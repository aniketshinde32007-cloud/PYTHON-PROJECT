# Objective Question 1: Average Lead Time for Cancelled vs Non-Cancelled Bookings

# Objective Question 4: Correlation Between Lead Time and Length of Stay

# Subjective Question 4: Impact of Day of the Week on Booking Cancellations

### Why are we doing this analysis?

We want to understand whether the day of the week on which a booking arrives has any impact on booking cancellations.

This helps us identify which days have higher or lower cancellation rates.

####Step 1 — Create arrival date

We need the actual arrival date first.

````python
df['arrival_date'] = pd.to_datetime(
    df['arrival_date_year'].astype(str) + '-' +
    df['arrival_date_month'] + '-' +
    df['arrival_date_day_of_month'].astype(str)
)
````

####Step 2 — Find cancellation rate by day

````python
df['day_of_week'] = df['arrival_date'].dt.day_name()

cancellation_by_day = (
    df.groupby('day_of_week')['is_canceled']
    .mean()
    .mul(100)
    .round(2)
)

cancellation_by_day
````

### Observation

The analysis shows that **Friday has the highest cancellation rate at 30.54%**, followed by Saturday at 30.10%. **Tuesday has the lowest cancellation rate at 24.32%**. This indicates that cancellations are slightly higher for bookings arriving toward the weekend, while Tuesday has the lowest cancellation rate.

## Visualization 2: Monthly Booking Trend

### Why are we making this graph?

We want to understand how hotel bookings change across different months of the year.

This helps us identify seasonal booking patterns and understand which months have higher or lower customer demand.

Since we want to compare the number of bookings across different months, we will use a countplot.

Customer segmentation means dividing customers into different groups based on their characteristics.

## Visualization 1: Bookings by Customer Type

### Why are we making this graph?

We want to understand how hotel bookings are distributed across different customer types.

Since we want to count the number of bookings in each customer type, we will use a countplot.

````python
plt.figure(figsize=(8,5))

sns.countplot(x='customer_type', data=df)

plt.title("Bookings by Customer Type")
plt.xlabel("Customer Type")
plt.ylabel("Number of Bookings")
plt.xticks(rotation=45)

plt.show()
````

### Observation

The graph shows that **Transient customers have the highest number of bookings**, while **Group customers have the lowest number of bookings**. ## Visualization 2: Cancellation Rate by Customer Type

### Why are we making this graph?

We want to understand how cancellation rates differ across different customer types.

This helps the hotel identify which customer segments are more likely to cancel their bookings.

Since we want to compare the cancellation rate between customer types, we will calculate the average of the `is_canceled` column for each customer type.

````python
cancellation_by_customer = (
    df.groupby('customer_type')['is_canceled']
    .mean()
    .mul(100)
    .round(2)
)

cancellation_by_customer
````

### Observation

The analysis shows that **Transient customers have the highest cancellation rate at 30.11%**, while **Group customers have the lowest cancellation rate at 9.93%**. Contract customers have a cancellation rate of 16.31%, while Transient-Party customers have a cancellation rate of 15.24%. This indicates that Transient customers are more likely to cancel their bookings compared with the other customer types.

# Actionable Insights and Recommendations

### Why are we doing this?

The purpose of this section is to convert the analysis findings into practical recommendations for hotel management.

The insights from booking patterns, cancellations, customer segments, pricing, and guest behavior can help the hotel improve revenue, reduce cancellations, and improve customer satisfaction.

### 1. Reduce High Cancellation Risk

Bookings with longer lead times showed a higher cancellation rate. Cancelled bookings had an average lead time of 105.72 days compared with 70.11 days for non-cancelled bookings.

**Recommendation:** The hotel can introduce flexible but controlled booking policies, reminders, and small incentives for guests who confirm their reservations closer to the arrival date.

### 2. Focus on Transient Customers

Transient customers represent the largest customer segment and also have the highest cancellation rate at 30.11%.

**Recommendation:** The hotel should focus targeted offers and confirmation reminders on Transient customers to reduce cancellations while maintaining this important customer segment.

### 3. Monitor Non-Refundable Bookings

Non-Refund bookings showed a very high cancellation rate of 94.70%.

**Recommendation:** The hotel should investigate this unusual pattern and review its non-refundable booking policy, payment process, and data quality before making major policy changes.

### 4. Improve High-Risk Distribution Channels

The Undefined distribution channel had an 80.00% cancellation rate, while TA/TO had a cancellation rate of 30.97%.

**Recommendation:** The hotel should investigate these channels, improve booking tracking, and work with high-cancellation channels to reduce cancellations.

### 5. Use Seasonal Pricing

August had the highest number of bookings and the highest average ADR, while January had fewer bookings.

**Recommendation:** The hotel can use higher pricing during high-demand periods and attractive offers during low-demand periods to improve occupancy and revenue.

### 6. Target International Guests

International guests accounted for a large share of bookings. Great Britain, France, and Spain were among the most common international markets.

International guests also had a lower cancellation rate of 23.75% compared with 35.67% for domestic guests.

**Recommendation:** The hotel can strengthen marketing campaigns in major international markets and create packages suited to international travellers.

### 7. Use Special Requests as a Customer Signal

Bookings with more special requests generally showed lower cancellation rates.

**Recommendation:** The hotel can use special requests as an engagement signal and provide personalized communication or services to improve customer satisfaction and encourage confirmed stays.

The analysis shows that hotel booking behavior is influenced by factors such as lead time, customer type, deposit type, distribution channel, seasonality, and guest origin.

The hotel can improve its operations by focusing on high-risk cancellation groups, strengthening international marketing, using seasonal pricing strategies, and improving booking-channel management.

These actions can help reduce cancellations, improve occupancy, increase revenue, and improve overall customer satisfaction.
